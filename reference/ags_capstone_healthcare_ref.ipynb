
# 📚 Step 2 – Import Libraries


```python

import os
import pandas as pd
import numpy as np

from transformers import pipeline

from langchain.memory import ConversationBufferMemory
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_community.document_loaders import PyPDFLoader
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.prompts import PromptTemplate
```

# 📂 Step 3 – Upload Medical Documents

Examples:
- WHO guidelines
- hospital documents
- medical PDFs


```python

from google.colab import files

uploaded = files.upload()
```


# 📄 Step 4 – Load PDF Documents


```python

loader = PyPDFLoader("medical_document.pdf")

documents = loader.load()

print("Total pages:", len(documents))
```


# 🧩 Step 5 – Text Chunking

Chunking improves:
- semantic retrieval,
- context precision,
- retrieval quality.


```python

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

texts = text_splitter.split_documents(documents)

print("Total chunks:", len(texts))
```


# 🧠 Step 6 – Create Embeddings


```python

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```


# 🗄️ Step 7 – Create Vector Database


```python

vector_db = FAISS.from_documents(
    texts,
    embedding_model
)

print("Vector database created.")
```


# 🔍 Step 8 – Semantic Search


```python

query = "What are kidney disease treatment methods?"

docs = vector_db.similarity_search(query)

docs[0]
```


# 🧠 Step 9 – Prompt Engineering


```python

prompt_template = PromptTemplate(
    input_variables=["context", "question"],
    template='''
You are a healthcare AI assistant.

Use ONLY the provided context.

Context:
{context}

Question:
{question}

Answer:
'''
)
```


# 🤖 Step 10 – Load LLM


```python

generator = pipeline(
    "text2text-generation",
    model="google/flan-t5-base"
)
```


# 🧠 Step 11 – Add Memory


```python

memory = ConversationBufferMemory(
    memory_key="chat_history",
    return_messages=True
)

print("Memory initialized.")
```


# 🧠 Step 12 – Simulate Agent Planning

Example Query:
"My father has kidney disease. Book a nephrologist appointment."

Break into subtasks:
- identify disease,
- retrieve treatment,
- specialist booking,
- summarize response.


```python

tasks = [
    "Identify disease",
    "Retrieve medical context",
    "Find specialist",
    "Generate response"
]

tasks
```


# 🛠️ Step 13 – Simulate Tool Calling


```python

def book_appointment(specialist):
    return f"Appointment request created for {specialist}"

response = book_appointment("Nephrologist")

print(response)
```


# 🧠 Step 14 – Build Context-Aware Prompt


```python

context = docs[0].page_content

question = "Summarize kidney disease treatment methods."

final_prompt = prompt_template.format(
    context=context,
    question=question
)

print(final_prompt)
```


# 🚀 Step 15 – Generate Response


```python

result = generator(
    final_prompt,
    max_length=200
)

print(result[0]['generated_text'])
```


# 📊 Step 16 – Evaluate Responses
